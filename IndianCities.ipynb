```python
import pandas as pd
import numpy as np
```

```python
Indian_Cities = pd.read_csv('cities_r2.csv')
Indian_Cities.head() # Display the first few rows of the DataFrame
```

```python

Indian_Cities.info() # Display information about the DataFrame
Indian_Cities.describe() # Display summary statistics of the DataFrame
Indian_Cities.isnull().sum() # Check for missing values in the DataFrame
```

```python
from matplotlib import pyplot as plt
import seaborn as sns
%matplotlib inline

# Select only numeric columns for correlation
corr = Indian_Cities.select_dtypes(include=[np.number]).corr() # Calculate correlation matrix
f, ax = plt.subplots(figsize=(10, 10))
sns.heatmap(corr, vmax=0.8, square=True) # Plot the correlation matrix
```

```python
fig = plt.figure(figsize=(20,10)) 
states = Indian_Cities.groupby('state_name')['name_of_city'].count().sort_values(ascending=True)
states.plot(kind="barh", fontsize=20, color='green')
plt.grid(visible=True, which='both', color='black')
plt.xlabel('No of cities taken from each state', fontsize=20)
plt.show()
```

```python
#List of states with the number of cities in each state
States = Indian_Cities.groupby('state_name')['name_of_city'].count().sort_values(ascending=True)
print(States)
```

```python
# change column name from 'name_of_city' to 'City' and 'state_name' to 'State'
Indian_Cities.rename(columns={'name_of_city': 'City', 'state_name': 'State'}, inplace=True)

print(Indian_Cities.head())
```
